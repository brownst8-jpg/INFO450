## 4. Conditionals (if / elif / else)  ⏱️ ~5 min

Conditionals let your code make decisions.

```python
if rating >= 9.0:

    print("Great")

    print("Good")
```

**Your turn:** Loop through the movies. For each one, print the title and a label:
- rating **9.0 or higher** → `"Masterpiece"`
- rating **8.0 to 8.9** → `"Great"`
- anything lower → `"Good"`

````python
# Loop through movies and print each title with its label.
# YOUR CODE HERE
for movie in movies:
    rating = movie["rating"]
    if rating >= 9.0:
        print(movie["title"], "Masterpiece")
    elif rating >= 8.0:
        print(movie["title"], "Great")
    else:
        print(movie["title"], "Good")
````

## 6. Functions  ⏱️ ~6 min

A **function** is reusable code you can call with different inputs. Define it with `def`, and `return` a result.

```python
def double(x):
    return x * 2
```

**Your turn:** Finish the function so it returns a nice description string for a movie dictionary.

````python
def describe_movie(movie):
    # Should return a string like:
    # "The Matrix is a Sci-Fi film from 1999 rated 8.7"
    # Use the movie's "title", "genre", "year", and "rating" keys.
    return f"{movie['title']} is a {movie['genre']} film from {movie['year']} rated {movie['rating']}"
    pass


# Test it (this should print a full sentence, not "None"):
print(describe_movie(movies[0]))
print(describe_movie(movies[2]))
````

**Your turn (part 2):** Write a function `average_rating` that takes the `movies` list and returns the average of all their ratings.

````python
def average_rating(movie_list):
    # Add up every movie's rating, then divide by how many there are.
    # Hint: len(movie_list) tells you how many movies there are.
    total = 0
    for movie in movie_list:
        total += movie["rating"]
    return total / len(movie_list)
    pass


print("Average rating:", average_rating(movies))
````

## 7. Lambda Functions  ⏱️ ~4 min

A **lambda** is a tiny one-line function with no name. These two are equivalent:

```python
def add_one(x): return x + 1
add_one = lambda x: x + 1
```

Lambdas are handy for **sorting**. `sorted()` takes a `key=` telling it what to sort by.

**Your turn:** Sort the movies by rating, highest first.

````python
# Sort movies by rating, HIGHEST first (descending).
# Fill in the lambda so it returns each movie's rating.
movies_by_rating = sorted(movies, key=lambda m: m["rating"], reverse=True)

for m in movies_by_rating:
    print(m["rating"], m["title"])
````
